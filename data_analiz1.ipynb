```
import pandas as pd
import numpy as np

np.random.seed(42)
data = pd.DataFrame({
    "EmployeeID": np.arange(1, 31),
    "Name": [f"Emp{i}" for i in range(1, 31)],
    "Department": np.random.choice(["HR", "Sales", "IT", "Finance"], size=30),
    "JoiningDate": pd.to_datetime(np.random.choice(pd.date_range("2018-01-01", "2023-01-01"), size=30)),
    "Salary": np.random.randint(40000, 120000, size=30),
    "Bonus": np.random.randint(2000, 15000, size=30),
    "ExperienceYears": np.random.randint(1, 15, size=30),
    "ProjectsCompleted": np.random.randint(0, 30, size=30),
    "RemoteDays": np.random.randint(0, 50, size=30)
})
```
