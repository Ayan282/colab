```
import pandas as pd
import numpy as np

# Sample sales data
sales_data = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct"],
    "Product_A": [250, 300, 400, 350, 500, 450, 480, 470, 510, 550],
    "Product_B": [200, 220, 330, 300, 360, 390, 400, 420, 450, 480],
    "Product_C": [150, 180, 210, 200, 230, 250, 260, 280, 300, 320]
})

# Sample employee data
employee_data = pd.DataFrame({
    "EmployeeID": np.arange(1, 16),
    "Name": [f"Emp{i}" for i in range(1, 16)],
    "Department": np.random.choice(["HR", "Sales", "IT", "Finance"], size=15),
    "Salary": np.random.randint(40000, 120000, size=15),
    "ProjectsCompleted": np.random.randint(0, 20, size=15),
    "ExperienceYears": np.random.randint(1, 15, size=15)
})
```
