```
import pandas as pd
import numpy as np

# Sample sales data
sales_data = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct"],
    "Product_A": [250, 300, 400, 350, 500, 450, 480, 470, 510, 550],
    "Product_B": [200, 220, 330, 300, 360, 390, 400, 420, 450, 480],
    "Product_C": [150, 180, 210, 200, 230, 250, 260, 280, 300, 320]
})

# Sample employee data
employee_data = pd.DataFrame({
    "EmployeeID": np.arange(1, 16),
    "Name": [f"Emp{i}" for i in range(1, 16)],
    "Department": np.random.choice(["HR", "Sales", "IT", "Finance"], size=15),
    "Salary": np.random.randint(40000, 120000, size=15),
    "ProjectsCompleted": np.random.randint(0, 20, size=15),
    "ExperienceYears": np.random.randint(1, 15, size=15)
})
```

```
sd["Umumi satis"]=sd["Product_A"]+sd["Product_B"]+sd["Product_C"]
sd["payA"]=sd["Product_A"]/sd["Umumi satis"]*100
sd["payB"]=sd["Product_B"]/sd["Umumi satis"]*100
sd["payC"]=sd["Product_C"]/sd["Umumi satis"]*100
sd
```

```
plt.bar(sd["Month"],sd["payA"],label="Product_A")
plt.bar(sd["Month"],sd["payB"],bottom=sd["payA"],label="Product_B")
plt.bar(sd["Month"],sd["payC"],bottom=sd["payA"]+sd["payB"],label="Product_C")
plt.legend()
plt.show()
```

```
sd["cumsumA"]=sd["Product_A"].cumsum()
sd["cumsumB"]=sd["Product_B"].cumsum()
sd["cumsumC"]=sd["Product_C"].cumsum()
sd
```

```
plt.plot(sd["cumsumA"],sd["Month"])
plt.plot(sd["cumsumB"],sd["Month"])
plt.plot(sd["cumsumC"],sd["Month"])
plt.fill_between(sd["cumsumA"],sd["Month"],color="blue",alpha=0.5)
plt.fill_between(sd["cumsumB"],sd["Month"],color="orange",alpha=0.5)
plt.fill_between(sd["cumsumC"],sd["Month"],color="green",alpha=0.5)
```
