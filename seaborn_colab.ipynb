```
import seaborn as sns
```

```
import pandas as pd
import numpy as np
import seaborn as sns

# Employee dataset
np.random.seed(42)
employee_data = pd.DataFrame({
    "EmployeeID": np.arange(1, 21),
    "Name": [f"Emp{i}" for i in range(1, 21)],
    "Department": np.random.choice(["HR", "Sales", "IT", "Finance"], size=20),
    "Salary": np.random.randint(40000, 120000, size=20),
    "ProjectsCompleted": np.random.randint(0, 20, size=20),
    "ExperienceYears": np.random.randint(1, 15, size=20),
    "RemoteDays": np.random.randint(0, 50, size=20)
})

# Sales dataset
sales_data = pd.DataFrame({
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct"],
    "Product_A": [250,300,400,350,500,450,480,470,510,550],
    "Product_B": [200,220,330,300,360,390,400,420,450,480],
    "Product_C": [150,180,210,200,230,250,260,280,300,320]
})
```

```
sd=pd.DataFrame(sales_data)
ed=pd.DataFrame(employee_data)
```

```
import matplotlib.pyplot as plt
```

```
sns.histplot(data=ed,x="Salary",bins=5)
plt.ylabel("Insan sayi")
```

```
sns.boxplot(x="Department",y="Salary",data=ed)
```

```
sns.scatterplot(data=ed,x="ExperienceYears",y="Salary")
```

```
sns.barplot(data=ed,x="Department",y="ProjectsCompleted")
```

```
sns.lineplot(data=sd,x="Month",y="Product_A")
```

```
sns.lineplot(data=sd,x="Month",y="Product_A",label="A")
sns.lineplot(data=sd,x="Month",y="Product_B",label="B")
sns.lineplot(data=sd,x="Month",y="Product_C",label="C")
plt.legend()
```

```
sns.barplot(x="Month",y="Product_A",label="A",data=sd)
```

```
sns.barplot(x="Month",y="Product_A",label="A",data=sd)
sns.barplot(x="Month",y="Product_B",label="B",data=sd)
sns.barplot(x="Month",y="Product_C",label="C",data=sd)

```

```
sns.boxplot(x="Department", y="Salary", hue="RemoteDays",data=ed)
plt.title("Salary Distribution by Department with RemoteDays Hue")
plt.show()
```

```
sns.violinplot(x="Department",y="Salary",data=ed,color="red")
```

```
sns.scatterplot(x="ExperienceYears",y="ProjectsCompleted",hue="Department",size="Salary",data=ed)
```

```
a=sns.FacetGrid(ed,col="Department")
a.map(sns.scatterplot,"ExperienceYears","Salary")
```

```
cols=["Salary", "ProjectsCompleted", "ExperienceYears", "RemoteDays"]
corr=ed[cols].corr()
sns.heatmap(ed[cols].corr())
```

```
sns.regplot(y="Salary", x="ExperienceYears",data=ed)
```

```
sns.jointplot(data=employee_data,x="ProjectsCompleted",y="Salary",kind="scatter")
```

```
sns.pairplot(ed,vars=["Salary","ExperienceYears","ProjectsCompleted","RemoteDays"])
```

```
sns.swarmplot(x='Department', y='Salary', data=ed)
```

```
sns.kdeplot(data=ed,x="Salary",hue="Department")
```
